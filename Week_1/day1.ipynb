```python
# Dynamic Typing:-
# Python is a dynamically typed language, which means that the type of a variable can change at runtime. We can assign a value of one type to a variable and then assign a value of another type to the same variable without any error.
# example:-
x = 10 # x is an integer
print(x) # this will print 10 on the screen 
x = "Hello" # now x is a string
print(x) # this will print Hello on the screen

# x is the same variable, but it can hold different types of data at different times in python. This is one of the features of python that makes it a dynamically typed language.

#  Static Typing:-
# In contrast to dynamic typing, static typing is a programming language feature where the type of a variable is determined at compile time and cannot change at runtime. In statically typed languages, you must declare the type of a variable before using it, and the type cannot be changed later in the program.
#example in C/C++/Java:
# int age = 25; // age is an integer variable
# age = "Hello"; // this will cause a compile-time error because age is declared as an integer variable and cannot hold a string value.
```

```python
# dynamic binding
# Dynamic binding, also known as late binding, is a programming language feature where the method or function to be called is determined at runtime based on the type of the object. In dynamically typed languages like Python, dynamic binding allows for more flexibility and polymorphism in code, as the same method can be called on different types of objects without needing to specify the type explicitly.
# example:
a= 5
print(a)
a="dheeraj"
print(a)

# Static binding, also known as early binding, is a programming language feature where the method or function to be called is determined at compile time based on the type of the object. In statically typed languages, static binding allows for faster execution and better performance, as the method to be called is determined before the program runs.
#example:
# int age = 25; // age is an integer variable
# System.out.println(age); // this will print 25 on the screen, and the method to be called is determined at compile time based on the type of the variable age, which is an integer.   
```

```python
# Stylish declaratiion techniques in python:-
# Unpacking in Python:-
a=1
b=2
c=3
print(a)
print(b)
print(c)
# output will be in different lines, because we have used three print statements to print the values of a, b, and c.
 # 1
 # 2
 # 3
 
a,b,c=1,2,3
print(a,b,c) 
# this will print 1 2 3 on the screen,this is called unpacking in python, where we can assign multiple values to multiple variables in a single line of code. The values on the right side of the assignment operator are unpacked and assigned to the variables on the left side in the same order.
```

### Type Conversion

```python
# There are two types of type conversion is python that is implicit type conversion and explicit type conversion.
# implicit type conversion:-  also known as coersion is type of conversion where python automatically converts one data type to another data type when we perform an operation on two different data types. For example, if we add an integer and a float, python will automatically convert the integer to a float before performing the addition.

# example of implicit type conversion:  

print(5+5.6) # this will print 10.6 on the screen, because python automatically converts the integer 5 to a float before performing the addition.

# problem with implicit type conversion:-

# print("The result is: " + 5) 

# this will cause a TypeError, because we are trying to concatenate a string and an integer, and python does not know how to convert the integer to a string in this context. To fix this, we can explicitly convert the integer to a string using the str() function:

# explicit type conversion:- also known as type casting is a type of conversion where we explicitly convert one data type to another data type using built-in functions like int(), float(), str(), etc. This is useful when we want to perform operations on different data types or when we want to display the output in a specific format.

print("The result is: " + str(5)) # this will print The result is: 5 on the screen, because we have explicitly converted the integer 5 to a string using the str() function before concatenating it with the other string.
```

## 7. Literals

```python
# literals:-# literals in python are basically fixed values written directly in the code that represent constant data. They are used to represent values that do not change during the execution of the program. Examples of literals in python include numbers (integers and floats), strings, booleans, and special values like None.

# Integer literals
a = 100
b = -50

# Floating-point literals
c = 3.14
d = -0.005

# Complex number literals
e = 4 + 7j
f = -3j

print(e.imag, e.real)
print(f.imag, f.real)

print(a, b, c, d, e, f)

x = 0b1010 # Binary literal (10 in decimal)
print(x)

y= 0o310 # Octal literal (200 in decimal)
print(y)

z = 0x1A # Hexadecimal literal (26 in decimal)
print(z)

# Different string literals
string = 'Hello'      # Single-quoted
strings = "Python"     # Double-quoted
multiline_str = '''This is 
a multi-line string'''  # Triple-quoted
raw_str = r"C:\Users\Python"  # Raw string
unicode = u"\U0001f600"  # Unicode string (😀 emoji)

print(string)
print(strings)
print(multiline_str)
print(raw_str)
print(unicode)

# Boolean literals
a = True  # True --> 1
b = False # False --> 0

print(a, b)       # Output: True False
print(1 == True)  # Output: True
print(0 == False) # Output: True
print(True + 5)   # Output: 6 (1 + 5)
print(False + 7)  # Output: 7 (0 + 7)

# Collection Literals
# Python provides four different types of literal collections:

# List literals: [1, 2, 3]
# Tuple literals: (1, 2, 3)
# Dictionary literals: {"key": "value"}
# Set literals: {1, 2, 3}


Rank = ["First", "Second", "Third"]  # List
colors = ("Red", "Blue", "Green")  # Tuple
Class = { "Jai": 10, "Anaya": 12 }  # Dictionary
unique_num = {1, 2, 3}  # Set

print(Rank, colors, Class, unique_num)

# Special Literal:- Python contains one special literal (None). 'None' is used to define a null variable. If 'None' is compared with anything else other than a 'None', it will return false.
res = None
print(res)
print(res == None)  # Output: True
print(res == 0)     # Output: False
```
