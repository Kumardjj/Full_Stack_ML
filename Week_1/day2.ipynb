# Operators:- 



```python
# operators:-
# operators are special symbols that perform specific operations on one or more operands. They are used to manipulate data and variables in programming languages.
# operands:- 
# operands are the values or variables on which the operators perform their operations. They can be constants, variables, or expressions that evaluate to a value.
 
# Operators in python:-
# 1. Arithmetic operators: +, -, *, /, %, **, //
# Relational operators: ==, !=, >, <, >=, <=
# Logical operators: and, or, not 
# Assignment operators: =, +=, -=, *=, /=, %=, **=, //=
# Bitwise operators: &, |, ^, ~, <<, >>
# Membership operators: in, not in
# Identity operators: is, is not
```

```python
# arithmatic operators:-
# 1. + (addition) :- it is used to add two numbers or concatenate two strings.
# example:-
a = 5
b = 10  
c = a + b
print(c)  # Output: 15

x = "Hello"
y = "World" 
z = x + " " + y
print(z)  # Output: Hello World

# 2. - (subtraction) :- it is used to subtract one number from another.
a = 10
b = 5   
c = a - b
print(c)  # Output: 5
# 3. * (multiplication) :- it is used to multiply two numbers or repeat a string a specified number of times.
a = 5
b = 10
c = a * b
print(c)  # Output: 50
x = "Hello"
y = x * 3
print(y)  # Output: HelloHelloHello
# 4. / (division) :- it is used to divide one number by another.    
a = 10
b = 5   
c = a / b
d= 5.5/2
print(d)
e= 5/2.0
print(e)
print(c)  # Output: 2.0
# 5. % (modulus) :- it is used to find the remainder when one number is divided by another.
a = 10
b = 3
c = a % b
print(c)  # Output: 1
# 6. ** (exponentiation) :- it is used to raise a number to the power of another number.
a = 2
b = 3
c = a ** b
print(c)  # Output: 8
# 7. // (floor division) :- it is used to divide one number by another and return the largest integer less than or equal to the result.
a = 10
b = 3   
c = a // b
print(c)  # Output: 3
```

```python
# Relational operators:-
# 1. == (equal to) :- it is used to compare two values for equality.
a = 5       
b = 5
c = (a == b)    
print(c)  # Output: True
# 2. != (not equal to) :- it is used to compare two values for inequality.
a = 5
b = 10
c = (a != b)
print(c)  # Output: True
# 3. > (greater than) :- it is used to compare if one value is greater than another.
a = 10
b = 5
c = (a > b)
print(c)  # Output: True
# 4. < (less than) :- it is used to compare if one value is less than another.
a = 5
b = 10
c = (a < b)
print(c)  # Output: True
# 5. >= (greater than or equal to) :- it is used to compare if one value is greater than or equal to another.
a = 10
b = 10
c = (a >= b)
print(c)  # Output: True
# 6. <= (less than or equal to) :- it is used to compare if one value is less than or equal to another.
a = 5
b = 10
c = (a <= b)
print(c)  # Output: True
```

```python
# Logical operators:-
# 1. and :- it is used to combine two or more conditions and returns True if all conditions are true, otherwise it returns False.
a = 5
b = 10
c = 15  
if a < b and b < c:
    print("Both conditions are true")
else:
    print("At least one condition is false")
# 2. or :- it is used to combine two or more conditions and returns True if at least one condition is true, otherwise it returns False.
a = 5
b = 10
c = 15
if a < b or b > c:
    print("At least one condition is true")
else:
    print("Both conditions are false")
# 3. not :- it is used to negate a condition and returns True if the condition is false, and False if the condition is true.
a = 5
b = 10
c = 15
if not a > b:
    print("Condition is false")
else:
    print("Condition is true")

```

```python
# Assignment operators:-
# 1. = (assignment) :- it is used to assign a value to a variable.
a = 5   
print(a)  # Output: 5
# 2. += (add and assign) :- it is used to add a value to a variable and assign the result back to that variable.
a = 5
a += 10    
print(a)  # Output: 15
# 3. -= (subtract and assign) :- it is used to subtract a value from a variable and assign the result back to that variable.
a = 10
a -= 5
print(a)  # Output: 5
# 4. *= (multiply and assign) :- it is used to multiply a variable by a value and assign the result back to that variable.
a = 5
a *= 10
print(a)  # Output: 50
# 5. /= (divide and assign) :- it is used to divide a variable by a value and assign the result back to that variable.
a = 10
a /= 5
print(a)  # Output: 2.0
# 6. %= (modulus and assign) :- it is used to find the remainder when a variable is divided by a value and assign the result back to that variable.
a = 10
a %= 3
print(a)  # Output: 1
# 7. **= (exponentiation and assign) :- it is used to raise a variable to the power of a value and assign the result back to that variable.
a = 2
a **= 3
print(a)  # Output: 8
# 8. //= (floor division and assign) :- it is used to divide a variable by a value using floor division and assign the result back to that variable.
a = 10
a //= 3
print(a)  # Output: 3
```

```python
# Bitwise operators:-
# 1. & (bitwise AND) :- it is used to perform a bitwise AND operation between two integers. It compares each bit of the first operand to the corresponding bit of the second operand and returns a new integer where each bit is set to 1 if both bits are 1, otherwise it is set to 0.
a = 5  # In binary: 0101
b = 3  # In binary: 0011
c = a & b  # In binary: 0001
print(c)  # Output: 1
# ex:-
a = 12  # In binary: 1100
b = 5   # In binary: 0101
c = a & b  # In binary: 0100
print(c)  # Output: 4

# 2. | (bitwise OR) :- it is used to perform a bitwise OR operation between two integers. It compares each bit of the first operand to the corresponding bit of the second operand and returns a new integer where each bit is set to 1 if at least one of the bits is 1, otherwise it is set to 0.
a = 5  # In binary: 0101
b = 3  # In binary: 0011
c = a | b  # In binary: 0111
print(c)  # Output: 7
# ex:-
a = 12  # In binary: 1100
b = 5   # In binary: 0101
c = a | b  # In binary: 1101
print(c)  # Output: 13
# 3. ^ (bitwise XOR) :- it is used to perform a bitwise exclusive OR operation between two integers. It compares each bit of the first operand to the corresponding bit of the second operand and returns a new integer where each bit is set to 1 if the bits are different, otherwise it is set to 0.
a = 5  # In binary: 0101    
b = 3  # In binary: 0011
c = a ^ b  # In binary: 0110
print(c)  # Output: 6
# ex:-
a = 12  # In binary: 1100
b = 5   # In binary: 0101
c = a ^ b  # In binary: 1001
print(c)  # Output: 9
# 4. ~ (bitwise NOT) :- it is used to perform a bitwise NOT operation on an integer. It takes the binary representation of the integer and inverts all the bits, changing 1s to 0s and 0s to 1s.
a = 5  # In binary: 0101    
c = ~a  # In binary: 1010 (which is -6 in two's complement)
print(c)  # Output: -6
# ex:-
a = 12  # In binary: 1100
c = ~a  # In binary: 0011 (which is -13 in two's complement)
print(c)  # Output: -13
# 5. << (left shift) :- it is used to perform a bitwise left shift operation on an integer. It shifts the bits of the integer to the left by a specified number of positions, filling the rightmost bits with zeros.
a = 5  # In binary: 0101
c = a << 2  # In binary: 010100 (which is 20 in decimal)
print(c)  # Output: 20
# ex:-
a = 12  # In binary: 1100
c = a << 3  # In binary: 11000000 (which is 96 in decimal)
print(c)  # Output: 96
# 6. >> (right shift) :- it is used to perform a bitwise right shift operation on an integer. It shifts the bits of the integer to the right by a specified number of positions, filling the leftmost bits with zeros for positive numbers and with ones for negative numbers.
a = 5  # In binary: 0101
c = a >> 2  # In binary: 0001 (which is 1 in decimal)
print(c)  # Output: 1
# ex:-
a = 12  # In binary: 1100
c = a >> 2  # In binary: 0011 (which is 3 in decimal)
print(c)  # Output: 3
```

```python
#membership operators:-
# 1. in :- it is used to check if a value is present in a sequence (such as a list, tuple, or string). It returns True if the value is found in the sequence, and False otherwise.
my_list = [1, 2, 3, 4, 5]   
print(3 in my_list)  # Output: True
print(6 in my_list)  # Output: False

# 2. not in :- it is used to check if a value is not present in a sequence. It returns True if the value is not found in the sequence, and False otherwise.
my_list = [1, 2, 3, 4, 5]
print(3 not in my_list)  # Output: False
print(6 not in my_list)  # Output: True
```

```python
# Identity operators:-
# 1. is :- it is used to check if two variables refer to the same object in memory. It returns True if both variables point to the same object, and False otherwise.
a = [1, 2, 3]       
b = a
print(a is b)  # Output: True
# ex:-
a = [1, 2, 3]
b = [1, 2, 3]
print(a is b)  # Output: False

# 2. is not :- it is used to check if two variables do not refer to the same object in memory. It returns True if both variables point to different objects, and False otherwise.
a = [1, 2, 3]   
b = a
print(a is not b)  # Output: False
# ex:-
a = [1, 2, 3]
b = [1, 2, 3]
print(a is not b)  # Output: True
```

## Loops in Python:-

```python
# need for loops:-
# loops are used to execute a block of code repeatedly until a certain condition is met. They are essential for performing tasks that require repetition, such as iterating over a collection of items, performing calculations, or automating tasks. Loops help to reduce code redundancy and improve efficiency by allowing you to write a block of code once and execute it multiple times with different inputs or conditions.
# ex:- In a hotel website, there are many pages that have the same container for image storing and display data. Instead of writing the same code for each page, we can use a loop to iterate through the pages and display the images and data dynamically. This way, we can save time and effort while maintaining consistency across the website. 
```

#### while loop:-

```python
# while loop:-
# while loops are used to execute a block of code repeatedly as long as a specified condition is true. The syntax for a while loop is:
# while condition:
#     # code to be executed 

# ex:- 
i = 1
while i<11:
    print(11,"*",i,"=",11*i)
    i=i+1
```
