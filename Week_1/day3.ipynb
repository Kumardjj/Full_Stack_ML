```python
# pattern1 :- 
# square of stars
n= int(input("Enter a number: "))
for i in range(n):
    for j in range(n):
        print("*",end=" ")
    print()
```

```python

# pattern 2:-
# triangle of stars
n= int(input("Enter a number: "))
for i in range(n):
    for j in range(i+1):
        print("*",end=" ")
    print()
```

```python
n= int(input("Enter a number: "))
for i in range(n):
    for j in range(n-i):
        print("*",end=" ")
    print()
```

```python
# mirrored triangle of stars
n= int(input("Enter a number: "))
for i in range(n):
    for j in range(n-i-1):
        print("  ",end=" ")
    for k in range(i+1):
        print("*",end=" ")
    print()
```

```python
# equilateral triangle of stars.
n= int(input("Enter a number: "))
for i in range(1,n+1):
    for j in range(1,n-i+1):
        print("  ",end=" ")
    for k in range(1,2*i):
        print("*",end=" ")  
    print()
```

```python
# break statement:-
# it is used to exit the loop when a certain condition is met. real life example is wehn you are searching aperson in a database and you find that person then you can use break statement to exit the loop and continue with the rest of the code.

for i in range(1,10):
    if i == 5:
        break
    print(i)
```

```python

low= int(input("Enter the lower limit: "))
high= int(input("Enter the higher limit: "))
for i in range(low, high+1):
    for j in range(2,i):
        if i%j ==0:
            break
    else:
        print(i)
        
```

```python
# continue statement:-
# it is used to skip the current iteration of the loop and continue with the next iteration. It is used when we want to skip a certain condition in the loop and continue with the rest of the code.
# real life example of continue is when we use a website  and you just want to skip a product and continue with the rest of the products. In that case you can use continue statement to skip that product and continue with the rest of the products.

for i in range(1,10):
    if i==4:
        continue
    print(i)    
```

```python
# pass statement:-
# It is used when we want to skip a certain condition in the loop and continue with the rest of the code. It is also used when we want to create a placeholder for a function or a class that we have not implemented yet.
# it is basically a method to avoid getting an error when we have not implemented a function or a class yet.
for i in range(1,10):
    pass
```

## note:-
text = "hello"
text.upper()  # This creates "HELLO" but doesn't save it anywhere
print(text)   # Output: "hello" (the original is unchanged)

To "change" it, you must reassign the variable:
text = text.upper()  
print(text)   # Output: "HELLO" (the variable now points to a new object) 

When you run text = "hello", Python creates a string object "hello" in memory and sticks the label text onto it.
When you run text = text.upper(), here is exactly what happens step-by-step:
The Right Side First: Python looks at text.upper(). It sees that text currently points to "hello", creates a brand-new string "HELLO", and holds it in a temporary spot in memory.

The Reassignment: Python sees the = sign. It "peels" the label text off the old "hello" object and moves it onto the new "HELLO" object.

Cleanup: The original "hello" object now has no labels attached to it. Python’s Garbage Collector eventually notices this and deletes the old "hello" from memory to save space.
So, even though the name stays the same, it is pointing to a completely different location in your computer's memory.

```python
# relational operations:-

print('delhi' == 'mumbai')

print('delhi' != 'delhi')

# lexiographyically comaprison 
print('mumbai' > 'pune')

print('pune'> 'mumbai')

# Lexicographical comparison (dictionary order) compares strings character by character based on their ASCII/Unicode values.
# First Character Comparison: The code compares the first letter of each string: 'm' (lowercase m) and 'p' (lowercase p).
# ASCII Values: In ASCII, lowercase letters are ordered from 'a' to 'z'. 'm' comes before 'p', meaning 'm' has a lower ASCII value than 'p'.
# Result: Because 'm' is not greater than 'p', the comparison 'mumbai' > 'pune' is immediately determined to be False.
```
