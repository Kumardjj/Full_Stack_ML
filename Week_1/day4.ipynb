# Array vs Lists

```python
import array as arr

# Arrays vs Lists in Python

# Python lists are dynamic arrays that can hold elements of different types.
# They are flexible but may use more memory due to overhead.

example_list = [1, 2, 3, 'a', 'b']
print("List:", example_list)

# The 'array' module provides typed arrays, which are more memory-efficient for homogeneous data.
# Arrays require all elements to be of the same type.


# Create an array of integers ('i' type code)
example_array = arr.array('i', [1, 2, 3, 4, 5])
print("Array:", example_array)

# Key differences:
# - Lists: Mutable, can hold mixed types, slower for large homogeneous data.
# - Arrays: Fixed type, more efficient for numerical computations, less flexible.

# Example operations:
# Append to list
example_list.append(4)
print("List after append:", example_list)

# Append to array (must be same type)
example_array.append(6)
print("Array after append:", example_array)

#
example_array.append(8)
print("Array after append:", example_array)
```

```python
import sys

list_50 = list(range(50))
array_50 = arr.array('i', range(50))

print("List length:", len(list_50))
print("List object size:", sys.getsizeof(list_50))
print("Estimated total list memory with element objects:",
    sys.getsizeof(list_50) + sum(sys.getsizeof(x) for x in list_50))

print("Array length:", len(array_50))
print("Array object size:", sys.getsizeof(array_50))
print("Array buffer info:", array_50.buffer_info())
print("Array typecode:", array_50.typecode)
print("Array item size:", array_50.itemsize)

```

```python
# Lists store references to Python objects. Each element in a list is a pointer to a separate Python object, so the list itself is an array of pointers.
# This makes lists flexible (can hold mixed types), but memory usage includes
# both the list container and the individual object overhead.

print("List container size:", sys.getsizeof(list_50))
print("Total list memory including element objects:",
    sys.getsizeof(list_50) + sum(sys.getsizeof(x) for x in list_50))

# Arrays from the array module store raw C-style values in a contiguous buffer.
# All elements must have the same typecode, so there is no per-item Python object
# overhead. This is more memory-efficient for homogeneous numeric data.

print("Array object size:", sys.getsizeof(array_50))
print("Array item size:", array_50.itemsize)
print("Estimated array buffer memory:", len(array_50) * array_50.itemsize)

# Summary:
# - list: container of pointers to Python objects, flexible, more overhead.
# - array: contiguous typed buffer, less overhead for homogeneous data.
```

```python
import sys

# Dynamic Arrays and Memory Allocation in Python

# A dynamic array automatically grows in size as more elements are added.
# Python lists are dynamic arrays that allocate extra capacity to avoid frequent reallocations.

# Theory:
# - When a list is created, it allocates a fixed amount of memory (capacity).
# - As elements are added, if the capacity is exceeded, the list reallocates with more memory (typically ~1.25x or 1.5x the current size).
# - This strategy balances memory usage and performance.

# Let's observe this with the 'l' variable and understand memory allocation:


# Current list 'l' = [1, 2, 3]
print("Initial list l:", l)
print("List size (bytes):", sys.getsizeof(l))

# Check capacity by appending elements and monitoring size changes
print("\n--- Memory allocation as we append elements ---")

l.append(4)
print("list size (bytes):",sys.getsizeof(l))
l.append(5)
print("list size (bytes):",sys.getsizeof(l))
l.append(6)
print("list size (bytes):",sys.getsizeof(l))
l.append(6)
print("list size (bytes):",sys.getsizeof(l))
l.append(6)
print("list size (bytes):",sys.getsizeof(l))
l.append(6)
print("list size (bytes):",sys.getsizeof(l))
# Observation: Notice that size doesn't increase with every append.
# The list allocates extra capacity in blocks to minimize reallocation overhead.

# This is more efficient than allocating exactly what's needed each time,
# because reallocation involves copying all existing elements to a new memory location.

print("\n--- Why this matters ---")
print("Dynamic allocation allows O(1) amortized time complexity for append operations.")
print("Without it, every append would be O(n) due to memory reallocation and copying.")
```
