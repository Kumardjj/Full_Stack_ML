### List comprehension:-

```python
# sorted() is a FUNCTION that returns a new sorted list
original_list = [3, 1, 4, 2]
sorted_result = sorted(original_list)
print("sorted() returns:", sorted_result)
print("original_list unchanged:", original_list)

# sort() is a METHOD that modifies the list in-place and returns None
original_list = [3, 1, 4, 2]
sort_result = original_list.sort()
print("sort() returns:", sort_result)
print("original_list modified:", original_list)

# This is why:
# - sorted() creates a new list (functional approach)
# - sort() modifies existing list in-place (procedural approach)
```

```python
# List comprehension - concise way to create lists

# Basic syntax: [expression for item in iterable]

# Example 1: Create a list of squares
squares = [x**2 for x in range(1, 6)]
print("Squares:", squares)

# Example 2: Filter even numbers
evens = [x for x in l if x % 2 == 0]
print("Even numbers from l:", evens)

# Example 3: Apply operation with condition
result = [x*2 for x in l if x > 2]
print("Double values > 2:", result)

# Example 4: Nested list comprehension
matrix = [[i*j for j in range(1, 4)] for i in range(1, 4)]
print("Matrix:", matrix)

# Example 5: List comprehension vs loop
# Traditional loop
traditional = []
for item in l:
    traditional.append(item * 2)

# List comprehension (same result)
comprehension = [item * 2 for item in l]
print("Traditional:", traditional)
print("Comprehension:", comprehension)
```

```python
# Nested list vs nested list comprehension

# Creating a nested list manually (traditional way)
nested_list = []
for i in range(1, 4):
    inner_list = []
    for j in range(1, 4):
        inner_list.append(i * j)
    nested_list.append(inner_list)
print("Nested list (manual):", nested_list)


# Using nested list comprehension (concise way)
nested_comprehension = [[i * j for j in range(1, 4)] for i in range(1, 4)]
print("Nested comprehension:", nested_comprehension)

# Both produce the same result: [[1, 2, 3], [2, 4, 6], [3, 6, 9]]
```

```python
# Two ways to traverse the list `l`

# 1. itemwise
print("itemwise traversal:")
for item in l:
    print(item, end=" ")
print()

# 2. indexwise
print("indexwise traversal:")
for i in range(len(l)):
    print(l[i], end=" ")
print()
```

```python
# zip function theory:
# - zip(iter1, iter2, ...) pairs elements from each iterable by position.
# - it returns an iterator of tuples.
# - it stops at the shortest iterable.
# - use list() or a loop to consume the result.
# - zip(*zipped) can unzip the paired tuples back into separate iterables.

# zip function examples
l=[1,2,3,4,5,6,7]
b=[5,6,7]

print("l:", l)
print("b:", b)

pairs = list(zip(l, b))
print("zipped pairs:", pairs)

# zip can take more than two iterables
numbers = [10, 20, 30]
print("zip(l, b, numbers):", list(zip(l, b, numbers)))

# unzip using zip(*...)
unzipped = list(zip(*pairs))
print("unzipped:", unzipped)

# unzip using zip(* )
unzipped2= list(zip(*pairs))
print("unzipped", unzipped2)

#addition using zip.
for x, y in zip(l, b):
    print(x, "+", y, "=", x + y)

# using list comprehension
lst = [i+j for i,j in zip(l,b)]
print(lst)
```

## Disadvantages of Python Lists

Python lists are versatile and convenient, but they come with several limitations:

1. **Fixed Data Type Limitations**: Lists can store mixed types, which can lead to runtime errors if type checking isn't done properly.

2. **Memory Inefficiency**: Lists consume more memory compared to arrays (like NumPy arrays) because each element is a Python object with overhead.

3. **Slow Operations**: List operations are slower than array operations, especially for numerical computations.

4. **No Element-wise Operations**: Unlike NumPy arrays, lists don't support element-wise arithmetic operations (e.g., `list1 + list2` concatenates, it doesn't add elements).

5. **Lack of Type Checking**: Python lists don't enforce data types, making code prone to errors if not carefully validated.

6. **Poor Performance for Large Data**: For large datasets, lists are significantly slower than NumPy arrays or other specialized data structures.

7. **Limited Mathematical Operations**: Lists lack built-in methods for mathematical operations like matrix multiplication or statistical functions.

8. **Mutability Issues**: Lists are mutable, which can lead to unintended side effects if references are shared.

9. **No Broadcasting**: Unlike NumPy arrays, lists don't support broadcasting for operating on different-sized data structures.

10. **Inefficient Search**: Searching through a list is O(n) time complexity; hash-based structures would be more efficient.

For numerical and scientific computing, **NumPy arrays** are preferred over Python lists.

note:- 
Every time you stop and restart a Python script, the operating system allocates a new memory slot for your variables. The address will be different in every "session."

This happens because of how your Operating System (OS) and the Python Interpreter manage resources. When you run a script, it isn’t just "code floating in space"—it lives inside a Process.

Here is why that address changes every time you hit "Run":

1. Dynamic Memory Allocation
When a Python script starts, it asks the OS for a chunk of RAM. The OS looks at its current "map" of memory—which is constantly changing because other apps (like your browser or Spotify) are also using RAM—and finds an available empty slot.

Since the available slots are different every time you run the script, the base address for your program changes.

2. Address Space Layout Randomization (ASLR)
This is actually a security feature built into modern Operating Systems (Windows, macOS, Linux).

The Problem: In the past, if a hacker knew exactly where a specific piece of data lived in RAM (e.g., address 0x7fff123), they could craft an attack to overwrite it.

The Solution: ASLR randomly "shuffles" the memory locations where a program's data is stored every time it launches. This makes it much harder for malicious code to predict where your variables are.

3. The Python Interpreter (CPython)
In Python, the id() function returns the memory address of the object. When your script ends, the Python Interpreter shuts down and releases all that memory back to the OS.

Session A: Python gets a block starting at Address 1000. Your list l is created at Address 1050.

Session B: Python gets a block starting at Address 5000. Your list l is now created at Address 5050.

How to visualize it
Think of RAM like a hotel.

When you "Run" your script, you are checking into the hotel.

The front desk (the OS) gives you whatever room is clean and empty at that moment (e.g., Room 201).

When the script "Finishes," you check out.

The next day, you come back to the same hotel, but Room 201 is now occupied by a different guest (another app), so the front desk gives you Room 405.

Your list l is like your suitcase—it's the same suitcase, but it's sitting in a different room number (address) every time you visit.
