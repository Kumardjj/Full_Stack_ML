### List comprehension:-

```python
# List comprehension - concise way to create lists

# Basic syntax: [expression for item in iterable]

# Example 1: Create a list of squares
squares = [x**2 for x in range(1, 6)]
print("Squares:", squares)

# Example 2: Filter even numbers
evens = [x for x in l if x % 2 == 0]
print("Even numbers from l:", evens)

# Example 3: Apply operation with condition
result = [x*2 for x in l if x > 2]
print("Double values > 2:", result)

# Example 4: Nested list comprehension
matrix = [[i*j for j in range(1, 4)] for i in range(1, 4)]
print("Matrix:", matrix)

# Example 5: List comprehension vs loop
# Traditional loop
traditional = []
for item in l:
    traditional.append(item * 2)

# List comprehension (same result)
comprehension = [item * 2 for item in l]
print("Traditional:", traditional)
print("Comprehension:", comprehension)
```

```python
# Nested list vs nested list comprehension

# Creating a nested list manually (traditional way)
nested_list = []
for i in range(1, 4):
    inner_list = []
    for j in range(1, 4):
        inner_list.append(i * j)
    nested_list.append(inner_list)
print("Nested list (manual):", nested_list)


# Using nested list comprehension (concise way)
nested_comprehension = [[i * j for j in range(1, 4)] for i in range(1, 4)]
print("Nested comprehension:", nested_comprehension)

# Both produce the same result: [[1, 2, 3], [2, 4, 6], [3, 6, 9]]
```

```python
# Two ways to traverse the list `l`

# 1. itemwise
print("itemwise traversal:")
for item in l:
    print(item, end=" ")
print()

# 2. indexwise
print("indexwise traversal:")
for i in range(len(l)):
    print(l[i], end=" ")
print()
```

```python
# zip function theory:
# - zip(iter1, iter2, ...) pairs elements from each iterable by position.
# - it returns an iterator of tuples.
# - it stops at the shortest iterable.
# - use list() or a loop to consume the result.
# - zip(*zipped) can unzip the paired tuples back into separate iterables.

# zip function examples
l=[1,2,3,4,5,6,7]
b=[5,6,7]

print("l:", l)
print("b:", b)

pairs = list(zip(l, b))
print("zipped pairs:", pairs)

# zip can take more than two iterables
numbers = [10, 20, 30]
print("zip(l, b, numbers):", list(zip(l, b, numbers)))

# unzip using zip(*...)
unzipped = list(zip(*pairs))
print("unzipped:", unzipped)

# unzip using zip(* )
unzipped2= list(zip(*pairs))
print("unzipped", unzipped2)

#addition using zip.
for x, y in zip(l, b):
    print(x, "+", y, "=", x + y)

# using list comprehension
lst = [i+j for i,j in zip(l,b)]
print(lst)
```

## Disadvantages of Python Lists

Python lists are versatile and convenient, but they come with several limitations:

1. **Fixed Data Type Limitations**: Lists can store mixed types, which can lead to runtime errors if type checking isn't done properly.

2. **Memory Inefficiency**: Lists consume more memory compared to arrays (like NumPy arrays) because each element is a Python object with overhead.

3. **Slow Operations**: List operations are slower than array operations, especially for numerical computations.

4. **No Element-wise Operations**: Unlike NumPy arrays, lists don't support element-wise arithmetic operations (e.g., `list1 + list2` concatenates, it doesn't add elements).

5. **Lack of Type Checking**: Python lists don't enforce data types, making code prone to errors if not carefully validated.

6. **Poor Performance for Large Data**: For large datasets, lists are significantly slower than NumPy arrays or other specialized data structures.

7. **Limited Mathematical Operations**: Lists lack built-in methods for mathematical operations like matrix multiplication or statistical functions.

8. **Mutability Issues**: Lists are mutable, which can lead to unintended side effects if references are shared.

9. **No Broadcasting**: Unlike NumPy arrays, lists don't support broadcasting for operating on different-sized data structures.

10. **Inefficient Search**: Searching through a list is O(n) time complexity; hash-based structures would be more efficient.

For numerical and scientific computing, **NumPy arrays** are preferred over Python lists.
