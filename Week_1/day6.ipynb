Tuples:-

```python
### Tuples in Python

# Tuples are immutable sequences in Python, similar to lists but cannot be modified after creation. They are defined using parentheses `()` and can contain elements of different types.

# #### Key Characteristics:
# - **Immutable**: Elements cannot be added, removed, or changed.
# - **Ordered**: Elements maintain their order.
# - **Heterogeneous**: Can hold mixed data types.

#### Creating a Tuple:
# Empty tuple
empty_tuple = ()
print(empty_tuple)

# Tuple with elements
my_tuple = (1, 2, 3, 'hello', 4.5)
print(my_tuple)


# Single element tuple (note the comma)
single_tuple = (42,)
print(single_tuple)


#### Accessing Elements:
my_tuple = (10, 20, 30, 40)
print(my_tuple[0])  # Output: 10
print(my_tuple[1:3])  # Output: (20, 30)


#### Common Operations:
# - **Length**: `len(my_tuple)`
# - **Concatenation**: `tuple1 + tuple2`
# - **Repetition**: `my_tuple * 2`
# - **Membership**: `element in my_tuple`

# Tuples are useful for fixed data that shouldn't change, like coordinates or constants. Unlike lists, they are hashable and can be used as dictionary keys.
```

```python
# Demonstrate common tuple operations

# Length
print(f"Length of my_tuple: {len(my_tuple)}")

# Concatenation
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
concatenated = tuple1 + tuple2
print(f"Concatenation: {concatenated}")

# Repetition
repeated = my_tuple * 2
print(f"Repetition: {repeated}")

# Membership
print(f"20 in my_tuple:- {20 in my_tuple}")
print(f"50 in my_tuple:- {50 in my_tuple}")
```

```python
# len
my_tuple = (1, 2, 3, 'hello', 4.5)
len(my_tuple)

# max() and min()
nums = (5, 10, 15, 2)
print(max(nums)) # Output: 15
print(min(nums)) # Output: 2

# sum 
nums = (1, 2, 3, 4)
print(sum(nums)) # Output: 10

# sorted
tup = (3, 1, 4, 2)
print(sorted(tup)) # Output: [1, 2, 3, 4]

# tuple
#tuple(): A constructor used to convert other iterables (like a list or string) into a tuple
my_list = [1, 2, 3]
print(tuple(my_list)) # Output: (1, 2, 3)
```

```python
# 3. speed (both are nearly same but it has been seen that tuple has faster than list.)
import time

L= list(range(10000000))
T=tuple(range(10000000))
start= time.time()
for i in L:
    i=i*5

print("List time: ",time.time()-start)

start1=time.time()
for i in T:
    i=i*5
end1 =time.time()
print("tuple time: ", time.time()-start1)
```

```python
# 4. memory (tuple took less time than list)
import sys
L= list(range(1000))
T=tuple(range(1000))

print("list size", sys.getsizeof(L))
print("tuple size", sys.getsizeof(T))
```
