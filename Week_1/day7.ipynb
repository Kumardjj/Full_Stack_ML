# Dictionary:-

dictionaries in python ae basically collection of key-value pairs , used to store data values like a map,which , unlike other data types which hold only a single value as an element.
in some langauges it is known as map or assosiative array.
dict = {'name':'dheeraj', 'age':33 , 'gender' :'male'}

characteristics:-

mutable

indexing has no meaning

keys cant be duplicated

keys cant be mutable item

```python
# empty dictionary.
d = {} # -->empty dictionary
print(type(d))

# 1D dictionary
# mixed
d= {'name':'dheeraj','gender':'male'}
print(d)

#homo
d= {(1,2,3):'dheeraj','age':24}
print(d)

# 2-D dictionary
d= {
    "student":"dheeraj",
    "college":"abes",
    "subjects":{
        "DSA":40,
        "OS":30,
        "CN":50 }}

# using sequence and dict fucntion:-

d4 = dict([('name','dheeraj'),('age',24)])
d4   # dictionary using list of tuples

d5 = dict((('name','dheeraj'),('age',24)))
d5   # dictionary using tuple of tuples 
```

```python
# duplicate keys:-
d= {'name':'dheeraj','name':'male'}
d #-->print only last key value pair , duplicate keys are not allowed.

#mutable items as keys are not allowed.
d= {'name':'dheeraj',(1,2,3):10}
d

```

```python
# access items:-

#using loops:-
d= {'name':'dheeraj','gender':'male'}
for key,value in d.items():
    print(key,value)

# usign keys
print(d['name']) # --> dheeraj

# get 
d.get('name')
d.get('gender')

# access item in 2-d dict.
d= {
    "student":"dheeraj",
    "college":"abes",
    "subjects":{
        "DSA":40,
        "OS":30,
        "CN":50 }}
print(d['subjects']['DSA'])
d['subjects']['new']=80
d
```

```python
# adding key value pairs:-
d['age']=24
print(d)
```

Remove key value paiirs:-

```python
d={'name': 'dheeraj', 'gender': 'male', 'age': 24}
#pop
d.pop('age')
d

#poitem 
d.popitem()
d

d={'name': 'dheeraj', 'gender': 'male', 'age': 24}
del d['name']
d

#clear()
d.clear()
d

d = {
    "student":"dheeraj",
    "college":"abes",
    "subjects":{
        "DSA":40,
        "OS":30,
        "CN":50}
    }
del d['subjects']['CN']
d
```

Editing key-value pair:-

```python
d = {
    "student":"dheeraj",
    "college":"abes",
    "subjects":{
        "DSA":40,
        "OS":30,
        "CN":50}
    }
d['subjects']['CN'] =100
d
```

```python
# update 
d1 = {1:2,3:4,4:5}
d2 = {4:7,6:8}
d1.update(d2)
print(d1)
print(d2)
```

```python
# dictionary comprehension.
new ={i:i**2 for i in range(1,10)}
print(new)
```

```python
d = {'DSA': 40, 'OS': 30, 'CN': 100}
new ={ key:value*2 for key,value in d.items()}
new
```

```python
days = ["sunday","monday","tuesday"]
temp = [30.5,32.6,35]
new= {key:value for key,value in zip(days,temp)} 
new
```

```python
products = {'phone':10,"laptop":30,"charger":0}
zero_product ={key:value for key, value in products.items() if products[key] == 0}
zero_product
```

```python
# nested comprehension:-
nested = {i:{j:i*j for j in range(1,11)}  for i in range(2,5) }
nested
```
