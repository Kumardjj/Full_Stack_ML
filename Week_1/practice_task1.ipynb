Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.

```python
# You can find the number of dogs using the formula: 
# dogs = (total legs - (2* total heads))//2
# . Once you have the number of dogs, simply subtract it from the total heads to get the number of chickens.

# Key Logic:- 
# Linear Equations:
# (C + D = Heads)
# (2C + 4D = Legs)

# Constraint Checks:The total number of legs must be an even number.The total legs cannot be less than if all animals were chickens (2 *Heads) or more than if all were dogs (4 *Heads).

# Calculation Example: If you have 10 heads and 28 legs, the program calculates ((28 - 2 * 10) / 2 = 4) dogs, and (10 - 4 = 6) chickens.

def solve_heads_legs():
    try:
        heads = int(input("Enter total no. of heads"))
        legs = int(input("Enter total no. of legs"))

        # Check if a valid solution exists
        if legs %2 != 0 or legs < 2* heads or legs > 4* heads :
            print("No valid solution exist")
        else:
            dogs = (legs - (2* heads))//2
            chickens = heads - dogs 
            print(f"dogs: {dogs}")
            print(f"chickens: {chickens}")
    except ValueError:
        print("please enter valid integers")
solve_heads_legs()


```

Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

tn = a + (n-1)*d  (nth term of a AP series)

a--> first term

n --> no. of term

d --> common difference = (nth term - (n-1)th term).





```python
first_term = int(input('enter 1st term'))
second_term = int(input('enter 2nd term'))
n = int(input('enter the value of n'))
d = second_term - first_term
nterm = first_term + (n-1)*d
print(nterm)
```

Integer Interning

or more specifically in Python:

Small Integer Caching

Python internally caches integers from: -5 to 256

```python
a= 1
b= 1
print(a is b)
print(id(a))
print(id(b))

# Memory:
# 1  ---> already stored object

# a ----┐
#       ├──> 1
# b ----┘
```

```python
a= 257
b=257
print(a is b)

# Why false: - 
# Python tries to optimize memory and speed by keeping some commonly used integers already created in memory.
# This is called integer interning or small integer caching.

# Python pre-creates integers from:

# -5 to 256

# so when you create a = 257 
# and b = 257 they basically holds different addresses in memory.

print(id(a))
print(id(b))

```

```python
a= 100
b= 100
print(a is b)
print(id(a))
print(id(b))

# Memory:
# 100  ---> already stored object

# a ----┐
#       ├──> 100
# b ----┘
```
