2. Check if an array is sorted


```python
arr = [1,2,3,4,5,6,7,10,19]
flag = False 
for i in range(1,len(arr)):
    if arr[i] < arr[i-1]:
        flag = True
if(flag == True):
    print("not sorted")
else:
    print("sorted")
```

3. Find Min/Max in a given array


```python
L = [21,1,34,23,54,11,10]

# maximum
max = -1e9
for i in L:
    if(i > max):
        max = i
print(max)

# min
min = 1e9
for i in L:
    if(i < min):
        min = i
print(min)
```

4. Find the first element to occur k times in an array


```python
l = [1,2,3,3,4,4,5,5]
k = 2
dic={}
for i in l:
    if i not in dic:
        dic[i] = 1
    else:
        dic[i] = dic[i] +1
for i in dic:
    if(dic[i] >= 2):
        print(i)
        break
```

5. Find duplicates in an array


```python
L = [1,1,2,3,4,4,5,5]
dup ={}
for i in L:
    if i not in dup:
        dup[i] = 1
    else:
        dup[i] = dup[i] + 1
for i in dup:
    if(dup[i]>1):
        print(i)
```
