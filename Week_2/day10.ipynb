```python
class atm:

    def __init__(self):
        # variables inside class
        self.pin = ""
        self.balance = 0
        self.menu()

    def menu(self):
        user_input = input("""
        hi how can i help you?
        1. Press 1 to create pin
        2. Press 2 to change pin
        3. Press  3 to check balance
        4. Press 4 to withdraw 
        5. Exit
              """)
        
        if user_input == "1":
            # create pin
            self.create_pin()

        elif user_input == "2":
            #change pin
            self.change_pin()

        elif user_input == "3":
            #check balance
            self.check_balance()
    
        elif user_input == "4":
            # withdrawl
            self.withdrawl()

        else:
            pass
    def create_pin(self):
        user_pin = input("Enter your pin")
        self.pin = user_pin

        user_balance = int(input("Enter your balance"))  
        self.balance = user_balance

        print("Pin created successfully")
        self.menu()

    def change_pin(self):
        old_pin= input("Enter your existing pin")
        if(old_pin == self.pin):
            # let him change the pin
            new_pin = input("Enter new pin")
            self.pin = new_pin
            print("Pin change successfully")
            self.menu()
        else:
            print('nahi karne de sakta re bidu')
            self.menu()
    def check_balance(self):
        user_pin = input("Enter your pin")
        if(user_pin == self.pin):
            print("Your balance is: ",self.balance)
        else:
            print('nahi bata sakta re baba')
    def withdrawl(self):
        user_pin = input("Enter your pin")
        if(user_pin == self.pin):
            amount = int(input("Enter the amount"))
            if(amount >= self.balance):
                self.balance = self.balance - amount 
                print("Withdrawl successful")
                print("Your current balance is",self.balance)
            else:
                print("paryapt balance nhi hai ")
        else:
            print("sale chor")
```

Constructor:-
In Python, a constructor is a special method used to initialize an object's state when it is created. While often referred to as a single entity, it technically involves two parts: creation and initialization.

Core Constructor Methods__init__(self) (The Initializer): 

This is what most developers call the "constructor". It is automatically invoked immediately after an object is created to set its initial attributes.

__new__(cls) (The Creator): This is the actual constructor that creates a new instance of the class and returns it. It is rarely overridden unless you are working with immutable types or singletons.Types of ConstructorsDefault 

Constructor: A simple constructor that takes no arguments except self. If you don't define one, Python provides a default that does nothing.

Parameterized Constructor: Accepts arguments to initialize an object with specific data at the time of creation.Example Syntaxpythonclass Student:
    # Parameterized constructor
    def __init__(self, name, grade):
        self.name = name   # Initializing attribute
        self.grade = grade # Initializing attribute
