# Functions:-

What are functions?
A function is a mathematical relationship where each input corresponds to exactly one output, often expressed as y = f(x). Acting as a "machine," it transforms independent variables into dependent outputs based on specific rules, with essential components including a domain (inputs) and a range (outputs).

Functions are reusable blocks of code designed to perform a specific task. Instead of writing the same instructions multiple times, you group them into a function, give it a name, and "call" it whenever needed.

What is a Function? (Non-Tech Analogy)Think of a function as a coffee machine:
Input (Arguments): You provide coffee beans and water.Process: The machine grinds and brews.
Output (Return Value): It produces a cup of coffee.
Reusability: You don't rebuild the machine every time you want coffee; you just press the "brew" button (call the function) again.

# *args and **kwargs:-

 *args and **kwargs are special python keywords that are used to pass the variable length of arguments to a fucntion.

*args :- used for normal arguments.

**kwargs :- used for keywords arguments

```python
# *args
def add(*a):
    return sum(a)
print(add(2,3,6,6))
```

```python
# multiply any number of arguments
def mul(*args):
    product =1
    for i in args:
        product=product*i
    print(args)
    return product

print(mul(2,3,4,5,6))
```
