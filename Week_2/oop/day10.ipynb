# self and need of self in oop in python:-


In Python Object-Oriented Programming (OOP), self is a conventional name for the first parameter of instance methods. 

It serves as a reference to the current instance of the class, allowing you to access and modify the object's specific data.

The Need for self:-

Without self, Python would not know which specific object's attributes or methods you want to interact with, especially when multiple objects exist from the same class. Its primary needs include:

Accessing Instance Attributes: It allows you to store and retrieve data unique to a specific object (e.g., self.name) rather than data shared by the whole class.

Differentiating Variables: It distinguishes between instance variables (attached to the object) and local variables defined within a method.

Method Interaction: It enables one method within a class to call another method on the same instance.

Implicit Passing: When you call a method like obj.method(), Python automatically converts it to Class.method(obj), passing the object itself as the self argument.


Key Characteristics:-

Not a Keyword: self is not a reserved word in Python; you could technically name it anything (like this), but using self is a strong community convention for readability.

Mandatory First Parameter: It must be the first parameter of any instance method, including the init constructor.

Memory Identity: Inside a class, self has the same memory address (id) as the object instance itself.Ex:-

```python
class Fractions:
    # parameterized constructor
    def __init__(self,x,y):
        self.num = x
        self.den = y

    def __str__(self):
        return '{}/{}'.format(self.num,self.den)
    
    def __add__(self,other):
        new_num = ((self.num * other.den) + (self.den * other.num))
        new_den = self.den * other.den
        return '{}/{}'.format(new_num,new_den)
    def __sub__(self,other):
        new_num = ((self.num * other.den) - (self.den * other.num))
        new_den = self.den * other.den
        return '{}/{}'.format(new_num,new_den)
    def __mul__(self,other):
        new_num = (self.num  * other.num)
        new_den = self.den * other.den
        return '{}/{}'.format(new_num,new_den)
    def __truediv__(self,other):
        new_num = (self.num  * other.den)
        new_den = self.den * other.num
        return '{}/{}'.format(new_num,new_den)
    
```

# Encapsulation:-

```python
# In above code when we write this:-
# sbi.  then we can see a list of data and methods of above class and the problem is that we we give access to our variable pin and balance to anyone then they can misuse it by change it.
# ex:-
sbi.balance = "Dheeraj"

# what we did is we change the balance to a string.
```

```python
# now we are calling the fucntion deposit.
sbi.deposit()  # --> generated error TypeError: can only concatenate str (not "int") to str.

# this happened because anyone can change the type of our variable balance.

# here comes the concept of Encapsulation.
```

The solution:-
solution is to make these variables private..
self.pin ----> self.__pin (private variable) 
so you can not access these variables putside the class.

```python
class atm:

    def __init__(self):
        # variables inside class
        self.__pin = ""
        self.__balance = 0
        self.menu()
        

    def menu(self):
        user_input = input("""
        hi how can i help you?
        1. Press 1 to create pin
        2. Press 2 to change pin
        3. Press 3 to check balance
        4. Press 4 to deposit 
        5. Press 5 to withdraw 
        6. Exit
              """)
        
        if user_input == "1":
            # create pin
            self.create_pin()

        elif user_input == "2":
            #change pin
            self.change_pin()

        elif user_input == "3":
            #check balance
            self.check_balance()
        
        elif user_input == "4":
            # deposit
            self.deposit()

        elif user_input == "5":
            # withdrawl
            self.withdrawl()
        
        else:
            pass

    def create_pin(self):
        user_pin = input("Enter your pin")
        self.__pin = user_pin

        user_balance = int(input("Enter your balance"))  
        self.__balance = user_balance

        print("Pin created successfully",self.__pin)
        self.menu()

    def change_pin(self):
        old_pin= input("Enter your existing pin")
        if(old_pin == self.__pin):
            # let him change the pin
            new_pin = input("Enter new pin")
            self.__pin = new_pin
            print("Pin change successfully")
            self.menu()
        else:
            print('nahi karne de sakta re bidu')
    def check_balance(self):
        user_pin = input("Enter your pin")
        if(user_pin == self.__pin):
            print("Your balance is: ",self.__balance)
        else:
            print('nahi bata sakta re baba')
    def deposit(self):
        user_pin = input("Enter your pin")
        if(user_pin == self.__pin):
            deposit= int(input("Enter deposit amount"))
            self.__balance = self.__balance + deposit
            print("current balance is: ", self.__balance)
        else:
            print("baap ko bhej tujhe pin nhi pata ")
            
    def withdrawl(self):
        user_pin = input("Enter your pin")
        if(user_pin == self.__pin):
            amount = int(input("Enter the amount"))
            if(amount >= self.__balance):
                self.__balance = self.__balance - amount 
                print("Withdrawl successful")
                print("Your current balance is",self.__balance)
            else:
                print("paryapt balance nhi hai ")
        else:
            print("sale chor")
```
